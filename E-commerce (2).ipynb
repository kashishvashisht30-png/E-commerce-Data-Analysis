# E-commerce database Analysis 

```python
import pandas as p

 #visualization
import plotly.express as px  #Data visualization library jo easy aur quick plots banata hai.
import plotly.graph_objects as go #Advanced and customizable graphs
import plotly.io as pio  #Graph templates ko customize karne ke liye
import plotly.colors as colors 
pio.templates.default ='plotly_white'  #Default theme white rakha gaya hai graphs ke liye
```

```python
df=p.read_csv("E-commerce.csv",encoding='latin-1')
```

```python
df
```

```python
df.info()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
df['order_date']=p.to_datetime(df['order_date'])
df['order_date'].dtype
```

```python
df['ship_date']=p.to_datetime(df['ship_date'])
df['ship_date'].dtype
```

```python
# create cols of monthly,yearly and weekly

df['order_month']=df['order_date'].dt.month
df['order_year']=df['order_date'].dt.year
df['Order_dayofweek']=df['order_date'].dt.dayofweek
```

```python
df.head(3)
```

### montly sales analysis

```python
sale_by_month=df.groupby('order_month')['sales'].sum().reset_index()

fig=px.line(sale_by_month,
            x='order_month', 
            y='sales',
            title='Monthly Sales Analysis')
fig.show()
sale_by_month
```

## category wise sales

```python
#Data Grouping:
#data.groupby('Order Month')['Sales'].sum() se har month ki total sales nikalte hain.
#.reset_index() data ko structured format me rakhta hai.
#px.line: Monthly sales trend show karne ke liye line chart banaya gaya hai.
#fig.show(): Graph display karta hai.
```

```python
sales_by_cat=df.groupby('category')['sales'].sum().reset_index()
sales_by_cat
```

```python
#groupby('Category'): Category-wise sales nikalte hain.
#Pie Chart:
#px.pie: Sales proportions ko pie chart me show karta hai.
#hole=0.5: Donut-style chart banata hai.
#Pastel Colors: Chart me soft color palette use kiya gaya hai.

sales_by_category = df.groupby('category')['sales'].sum().reset_index()


fig = px.pie(sales_by_category, 
             values='sales', 
             names='category', 
             hole=0.5, 
             color_discrete_sequence=px.colors.qualitative.Pastel)

fig.update_traces(textposition='inside', textinfo='percent+label')
fig.update_layout(title_text='Sales Analysis by Category', title_font=dict(size=24))

fig.show()
```

## analysis by sub category

```python
df.columns
```

```python


sales_by_subcategory = df.groupby('sub-category')['sales'].sum().reset_index()
fig = px.bar(sales_by_subcategory, 
             x='sub-category', 
             y='sales', 
             title='Sales Analysis by Sub-Category')
fig.show()
sales_by_subcategory
```

## monthly profit

```python
df.head(3)
```

```python
profit_by_month = df.groupby('order_month')['profit'].sum().reset_index()
fig = px.line(profit_by_month, 
              x='order_month', 
              y='profit', 
              title='Monthly Profit Analysis')
fig.show()
```

## profit by category


```python
profit_by_categ = df.groupby('category')['profit'].sum().reset_index()
fig = px.bar(profit_by_categ, 
              x='category', 
              y='profit', 
              title='Monthly Profit Analysis')
fig.show()
profit_by_categ
```

```python
fig=px.pie(profit_by_categ,
           values='profit',
           names='category',
           hole=0.5,
           color_discrete_sequence=px.colors.qualitative.Pastel)
fig.update_traces(textposition='inside', textinfo='percent+label')
fig.update_layout(title_text='Profit by category',title_font=dict(size=24))
fig.show()
```

## profit by sub_categ

```python
profit_by_subc=df.groupby('sub-category')['profit'].sum().reset_index()
profit_by_subc

fig=px.bar(profit_by_subc,
           x='sub-category',
           y='profit',
           title='Profit by Sub-Category')
fig.show()
```

```python
fig=px.pie(profit_by_subc,
           values='profit',
           names='sub-category',
           hole=0.5,
           title='Profit by Sub-Category')
fig.update_traces(textposition='inside',textinfo='percent+label')
fig.update_layout(title_text='Profit by sub-category',title_font=dict(size=24))
fig.show()
```

## sales and profit by customer segment

```python
sales_profit_by_segment = df.groupby('segment').agg({'sales': 'sum', 'profit': 'sum'}).reset_index()

color_palette = colors.qualitative.Pastel

fig = go.Figure()
fig.add_trace(go.Bar(x=sales_profit_by_segment['segment'], 
                     y=sales_profit_by_segment['sales'], 
                     name='sales',
                     marker_color=color_palette[0]))

fig.add_trace(go.Bar(x=sales_profit_by_segment['segment'], 
                     y=sales_profit_by_segment['profit'], 
                     name='profit',
                     marker_color=color_palette[1]))

fig.update_layout(title='Sales and Profit Analysis by Customer Segment',
                  xaxis_title='Customer Segment', yaxis_title='Amount')

fig.show()
```

## sales to profit ratio

```python

sales_profit_by_segment = df.groupby('segment').agg({'sales': 'sum', 'profit': 'sum'}).reset_index()
sales_profit_by_segment['Sales_to_Profit_Ratio'] = sales_profit_by_segment['sales'] / sales_profit_by_segment['profit']
print(sales_profit_by_segment[['segment', 'Sales_to_Profit_Ratio']])
```
